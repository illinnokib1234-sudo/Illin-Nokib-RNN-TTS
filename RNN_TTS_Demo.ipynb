```python
from pathlib import Path

# Create output folder
OUTPUT_DIR = Path("rnn_audio")
OUTPUT_DIR.mkdir(exist_ok=True)

# Slide-by-slide presentation script
slides = {
    1: """
Assalamu Alaikum. This is Illin Nokib.
Today, I will be presenting on Recurrent Neural Networks, or RNNs,
and their fundamental role in Natural Language Processing.
""",

    2: """
Unlike traditional feedforward neural networks that process inputs independently,
Recurrent Neural Networks are designed for sequential data,
where the order of information matters.
At each step, an RNN takes the current input
and combines it with information from the previous step.
This creates a hidden state, which acts like a form of memory
and allows the network to maintain context across a sequence.
""",

    3: """
Language is naturally sequential, variable, and context-dependent.
For example, the word bank can refer to a financial institution
or the edge of a river, depending on the surrounding words.
RNNs can handle sequences of different lengths
and reuse the same parameters at every time step.
This makes them suitable for processing language and other sequential data.
""",

    4: """
When we unroll the RNN loop across time,
we can see how information flows from t one to t four.
At every time step, the hidden state is updated
using the current input x t
and the previous hidden state h t minus one.
This continuous chain allows information from earlier parts
of the sequence to influence later parts.
""",

    5: """
However, standard RNNs can struggle with long sequences
because of the vanishing gradient problem.
Important information may gradually be lost as the sequence becomes longer.
Long Short-Term Memory networks, or LSTMs,
address this problem using a dedicated cell state
and three gates: the Forget gate, Input gate, and Output gate.
GRUs, or Gated Recurrent Units, provide a simpler alternative
using Reset and Update gates.
""",

    6: """
RNNs can be adapted to several different sequence-based tasks.
Many-to-Many architectures can be used for machine translation.
Many-to-One architectures can be used for sentiment analysis.
One-to-Many architectures can be used for text generation.
RNNs can also perform sequence labeling,
such as identifying named entities in text.
""",

    7: """
Although RNNs played an important role in the development of NLP,
modern NLP applications often use Transformers.
RNNs process words sequentially,
which can create a training bottleneck.
Transformers can process the sequence in parallel
using a mechanism called self-attention.
However, RNNs can still be useful when lightweight,
real-time, or on-device processing is important.
""",

    8: """
That concludes my presentation on Recurrent Neural Networks.
Thank you for your time.
"""
}

print(f"Prepared {len(slides)} slides.")
print("Speaker name: Illin Nokib")
print("Output folder:", OUTPUT_DIR)
```

```python
from pathlib import Path
from IPython.display import Audio, display

# Voice reference
REFERENCE = "my_voice_reference (1).wav"

# Generate one audio file for each slide
for slide_number, text in slides.items():

    output_file = OUTPUT_DIR / f"slide_{slide_number}.wav"

    print(f"Generating Slide {slide_number}...")

    tts.tts_to_file(
        text=text.strip(),
        speaker_wav=REFERENCE,
        language="en",
        file_path=str(output_file)
    )

    print(f"✓ Slide {slide_number} completed")

print("\nAll 8 slide audio files generated successfully! 🎉")
```
